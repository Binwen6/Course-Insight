```python
import numpy as np
from scipy.fftpack import fft

# 生成一个样本信号，例如一个正弦波
t = np.linspace(0, 1, 1400) # 时间向量
y = np.sin(2 * np.pi * 200 * t) # 生成一个200Hz的正弦波

# 执行快速傅里叶变换
fft_y = fft(y)

# 获取频率和振幅
N = 1400 # 采样点数
x = np.arange(N) # 频率个数
abs_y = np.abs(fft_y) # 振幅谱
angle_y = np.angle(fft_y) # 相位谱

# 由于FFT结果是对称的，只取一半区间
half_x = x[:N // 2]
normalization_y = abs_y / N # 归一化处理
normalization_half_y = normalization_y[:N // 2] # 取一半区间的振幅谱
```

```python
import matplotlib.pyplot as plt

# 处理中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 绘制原始信号
plt.figure()
plt.plot(t, y)
plt.title('原始信号')

# 绘制振幅谱
plt.figure()
plt.plot(half_x, normalization_half_y)
plt.title('振幅谱')

# 绘制相位谱
plt.figure()
plt.plot(half_x, angle_y[:N // 2])
plt.title('相位谱')

plt.show()
```
